```python
def get_full_text_cleaned(text_element):
    # Extract the 'data-html' attribute content and decode HTML entities
    html_content = html.unescape(text_element.get("data-html", ""))
    
    # Parse the HTML content
    soup = BeautifulSoup(html_content, "html.parser")
    
    # Use a set to track unique lines and maintain order
    seen = set()
    lines = []
    
    for p in soup.find_all("p"):
        text = p.get_text(separator=" ", strip=True)
        if text not in seen:
            seen.add(text)
            lines.append(text)
    
    # Join lines with newline characters
    formatted_text = " \n ".join(lines)
    
    return formatted_text
```

```python
def get_full_text(driver, writing_url):
    """Fetch the full text of a writing."""
    try:
        driver.get(writing_url)
        time.sleep(2)

        soup = BeautifulSoup(driver.page_source, "html.parser")
        # text_element = soup.select_one("div.full-text-selector")
        text_element = soup.select_one('#HtmlRawText')
        text = get_full_text_cleaned(text_element)
        words = text.split(' ')  # Split the string into words
        n = len(words)  # Get the total number of words
        print(' '.join(words[n//2:]))
        return ' '.join(words[n//2:])
    except:
        driver = webdriver.Chrome(service=service, options=chrome_options)
        driver.get(writing_url)
        time.sleep(2)

        soup = BeautifulSoup(driver.page_source, "html.parser")
        # text_element = soup.select_one("div.full-text-selector")
        text_element = soup.select_one('#HtmlRawText')
        text = get_full_text_cleaned(text_element)
        words = text.split(' ')  # Split the string into words
        n = len(words)  # Get the total number of words
        print(' '.join(words[n//2:]))
        return ' '.join(words[n//2:])
```

```python
def main():
    all_writings = {}
    
    db_init = pd.read_csv('shayari_db.csv')
    already_done = list(db_init['AUTHOR'].unique())

    # Loop through alphabet pages
    # for letter in "ABCDEFGHIJKLMNOPQRSTUVWXYZ":
    for letter in "D":
        letter_url = f"{BASE_URL}{letter}"
        print(f"Fetching authors for letter {letter}...")

        # If author_links NOT already read
        author_links = get_author_links(driver, letter_url)

        for i in range(len(author_links)):
            if author_links[i][:4] == "http":
                pass
            else:
                author_links[i] = "https://www.rekhta.org"+author_links[i]
        
        with open("link_list.txt", 'w') as f:
            for s in author_links:
                f.write(str(s) + '\n')

        # If author_links already read    
        # with open("link_list.txt", 'r') as f:
        #     author_links = [line.rstrip('\n') for line in f]
        
        for author_url in author_links:
            author_name = author_url.split('/')[-2]
            if author_name in already_done:
                continue
            else:
                already_done.append(author_name)
            if author_name in all_writings.keys():
                continue
            print(f"Fetching writings for author: {author_name}")
            writing_links = get_writing_links(driver, author_url)

            for i in range(len(writing_links)):
                if writing_links[i][:4] == "http":
                    pass
                else:
                    writing_links[i] = "https://www.rekhta.org"+writing_links[i]


            writings = []
            for writing_url in writing_links:
                text = get_full_text(driver, writing_url)
                writings.append(text)
                csv_filename = 'shayari_db.csv'
                with open(csv_filename, mode="a", newline="", encoding="utf-8") as file:
                    writer = csv.writer(file)
                    
                    # Write the author and formatted poetry
                    writer.writerow([author_name, text])
                time.sleep(1)  # Be polite by adding a delay between requests

            all_writings[author_name] = writings


    # Save or process the collected data
    for author, writings in all_writings.items():
        print(f"Author: {author}")
        for idx, text in enumerate(writings):
            print(f"  Writing {idx + 1}: {text[:100]}...")  # Preview first 100 characters

    driver.quit()
```

```python
if __name__ == "__main__":
    
    main()
```
